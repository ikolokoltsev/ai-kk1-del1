### 1. Hur hänger AI, ML och DL ihop?

AI is the broadest term. It covers everything where a computer does something that looks intelligent, including rule-based systems where a human wrote all the rules by hand.

ML is a part of AI. Instead of writing the rules ourselves, we give the computer data and it learns the patterns from it.

DL is a part of ML that uses neural networks with many layers. It can learn useful features from raw data by itself, instead of us doing the feature engineering.

So they are nested: DL is inside ML, and ML is inside AI.

### 2. Vilka är de fyra problemkategorierna inom ML?

1. Supervised learning: we have both X and the correct answer y, and the model learns to predict y from X. Divided into regression (y is a number) and classification (y is a category).
2. Unsupervised learning: we only have X, no answers. The model finds structure by itself, for example clustering similar customers or reducing the number of dimensions (PCA).
3. Semi-supervised learning: a mix, only a small part of the data has labels. Useful when labeling is expensive, like paying people to tag images.
4. Reinforcement learning: an agent acts in an environment and gets rewards or penalties, and learns over time which actions are good. Used for example in games and robotics.

### 3. Förklara följande:

#### a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och testdata?

- Training data: the model learns its parameters from this.
- Validation data: used to compare models and tune hyperparameters.
- Test data: used once at the end, to estimate how the model performs on data it has never seen.

The point is to get an honest estimate of how the model works on new data, not how well it remembers the data it trained on. If we tune things based on the test data, information from it leaks into the model and the test result becomes too optimistic.

#### b) Vad är k-delad korsvalidering för något?

We split the training data into k parts (folds) of about the same size. Then we train k times: each time one fold is used for validation and the other k-1 folds for training. At the end we average the k scores.

Every row is used for validation exactly once, so no data is wasted and the score is more stable than from a single split. The downside is that we have to train k times. Common choices are k=5 or k=10.

#### c) Vad är root mean squared error (RMSE) för något?

It's a way to measure how wrong the regression model's predictions are on average.

$$RMSE = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$$

You take the difference between real value and prediction, square it (so negatives don't cancel and big mistakes get punished harder), average all of that, then take square root at the end to get back to the original unit.

Lower RMSE = better model. If RMSE is 50000 on a house price model, that means the errors, squared then averaged then square-rooted, work out to 50k kr, roughly the typical size of the errors but not literally the average absolute error (that's what MAE is for, and it's usually a bit smaller than RMSE since RMSE weighs big errors more).

#### d) Vad är en hyperparameter för något? Vad är en parameter för något?

- Parameter: something the model learns from the training data, like the coefficients in linear regression or the weights in a neural network. `fit()` calculates them, we don't set them by hand.
- Hyperparameter: something we choose before training that controls how the model learns or how complex it is. For example `alpha` in Ridge, `k` in KNN or `max_depth` in a decision tree.

Short version: parameters are learned, hyperparameters are chosen by us, usually with the help of cross-validation or grid search.

#### e) Vad är grid search och hur hänger namnet “grid” och “search” ihop med processen som genomförs? Läser vi dokumentationen: https://scikit-learn.org/stable/modules/generated/sklearn.model_selection.GridSearchCV så ser vi att hyperparametern refit har standardvärdet True, vad innebär det?

Grid search is a way to find good hyperparameters by trying combinations of them.

"Grid": we give a list of values for each hyperparameter, e.g. `alpha = [0.1, 1, 10]` and `fit_intercept = [True, False]`. All combinations form a grid, here 3 x 2 = 6 points.

"Search": it goes through every point in the grid, trains and evaluates the model with cross-validation, and keeps track of which combination scored best.

`refit=True`: after the search, sklearn trains one final model with the best hyperparameters on all the data passed to `fit()` (not just k-1 folds). This model is saved as `best_estimator_`, so we can call `.predict()` directly on the GridSearchCV object. With `refit=False` we only get the search results, not a trained model.

#### f) Vad är kategorisk data och hur hanteras det? I ditt svar, använd begreppen nominal data, ordinal data, one-hot-encoding, dummy-variable-encoding och ordinal encoding.

Categorical data is data where the value is a category, not a number.

Two types:
- Nominal: no order between the categories, like colors or cities.
- Ordinal: there is a natural order, like small/medium/large or grades. The distance between the steps isn't necessarily equal though.

Most models need numbers, so the categories have to be encoded:
- Ordinal encoding: replace each category with an integer following the order (0, 1, 2...). Fits ordinal data. For nominal data it's a bad idea, since the model would think one city is "larger" than another.
- One-hot encoding: one new 0/1 column per category. Fits nominal data since no order is implied.
- Dummy variable encoding: like one-hot but one column is dropped, so k categories become k-1 columns. The dropped category is the reference (all zeros). This avoids perfect multicollinearity (the dummy variable trap), which matters for example for linear regression with an intercept.

#### g) Vad är feature engineering för något?

Feature engineering means using domain knowledge to create, transform or select features so the model can learn better. For example extracting weekday or month from a date, calculating BMI from weight and height, log-transforming a skewed variable like income, creating interaction features, or scaling.

Good features often improve results more than switching to a more advanced model.

#### h) Vad menas med principle of parsimony?

It's the same idea as Occam's razor: if two models perform about equally well, choose the simpler one.

A simpler model is easier to understand and explain, has less risk of overfitting, often generalizes better, and is cheaper and faster to train and run.

### 4. Vad menas med att “en modell är en förenkling av verkligheten”?

Reality depends on a huge number of factors, and many of them we can't measure or don't even know about. A model only uses the information we give it and tries to capture the main patterns. Everything else is treated as noise.

For example, a house price model might use size, location and number of rooms, but not things like how motivated the buyer was that day.

So a model will never be perfectly accurate, but it can still be useful ("all models are wrong, but some are useful"). If we try to make it capture everything, it starts fitting noise instead and overfits.

### 5. Vad menas med att en modell är “överanpassad” eller overfitted på engelska?

It means the model has learned the training data too well, including noise and random details that don't generalize. It has more or less memorized the data instead of learning the underlying pattern.

You can see it when the training error is low but the validation or test error is much higher.

Common causes are a too complex model, too little data or too many features. Ways to fix it: a simpler model, regularization (Ridge/Lasso), more data, fewer features, and using cross-validation to detect it.

The opposite is underfitting: the model is too simple and does badly on both training and test data.

### 6. Högre är bättre i scikit-learn scoring, vad innebär det?

In sklearn's model-selection tools (`cross_val_score`, `GridSearchCV`, etc), all "scorer" objects follow this convention: bigger score number always means better model, so these tools can always just maximize.

That's fine for accuracy or R² since higher is naturally better there. But for error-type metrics (MSE, MAE, RMSE) lower error is actually better, so the scorer versions used by these tools flip the sign, e.g. `neg_mean_squared_error`. So -20 counts as better than -50 in that context, even tho the actual MSE of 20 is smaller/better than 50. This only applies to the scorer strings/objects though, the raw metric functions like `mean_squared_error()` called directly still just return the plain error value, lower is better there, no sign flip.

### 7. Vad är tvärsnittsdata, tidsseriedata och paneldata? Exemplifiera när respektive datakategori kan uppstå.

- Cross-sectional data (tvärsnittsdata): many units observed at one point in time. For example a survey answered by 1000 people in 2025, or all apartments sold in Stockholm during one month.
- Time series data (tidsseriedata): one unit observed at many points in time. For example the daily closing price of a stock, or monthly unemployment in Sweden.
- Panel data (paneldata): many units observed at many points in time, a combination of the two. For example the income of the same 500 households every year for 10 years.

Tools like `GridSearchCV` and `cross_val_score` are written to always maximize the score. Instead of having separate logic for metrics we want to minimize, sklearn flips the sign of error metrics. Minimizing MSE is the same as maximizing -MSE:

- Model B: MSE = 50, score = -50

-20 is larger than -50, so model A wins, which is correct since it has the smaller error.
